# LearnTrack ML: Exploratory Data Analysis (EDA)
## Academic Performance Prediction Dataset
This notebook documents statistical distributions, feature correlations, and target characteristics for the LearnTrack regression pipeline.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/academic_performance.csv')
print(f'Dataset Loaded: {df.shape[0]} rows, {df.shape[1]} features')
df.head()
```

### 1. Data Integrity & Missing Value Verification
Verifying zero null entries across all 6 predictive features and target variable.

```python
print('Missing Values:\n', df.isnull().sum())
print('Duplicate Rows:', df.duplicated().sum())
```

### 2. Descriptive Summary Statistics
Examines central tendency, dispersion, and quartiles across features.

```python
df.describe().round(2)
```

### 3. Correlation Matrix Analysis
Evaluates linear Pearson correlation coefficients between academic features and final score.

```python
corr_matrix = df.corr().round(3)
print('Correlation with final_score:\n', corr_matrix['final_score'].sort_values(ascending=False))
corr_matrix
```

### 4. Grade Distribution
Categorical spread based on standard university thresholds (A+ to F).

```python
def assign_grade(score):
    if score >= 90: return 'A+'
    elif score >= 80: return 'A'
    elif score >= 70: return 'B+'
    elif score >= 60: return 'B'
    elif score >= 50: return 'C'
    elif score >= 40: return 'D'
    else: return 'F'

df['grade'] = df['final_score'].apply(assign_grade)
print(df['grade'].value_counts())
```
